# 02 - Data Exploration (EDA)

## Dataset path (clean folder)

```python
import os
from PIL import Image
import random
import matplotlib.pyplot as plt

BASE_PATH = os.path.join("..", "..", "datasets", "b_classification_flowers")
DATASET_PATH = os.path.join(BASE_PATH, "b_classification_flowers_clean")

class_names = sorted(name for name in os.listdir(DATASET_PATH) if os.path.isdir(os.path.join(DATASET_PATH, name)))
print(class_names)
```

## Number of images per class

```python
class_counts = {}

for class_name in class_names:
    class_path = os.path.join(DATASET_PATH, class_name)

    if os.path.isdir(class_path):
        images = [
            f for f in os.listdir(class_path)
            if f.lower().endswith(('.jpg', '.jpeg', '.png'))
        ]

        class_counts[class_name] = len(images)

plt.figure(figsize=(8, 5))
plt.bar(class_counts.keys(), class_counts.values())

plt.title("Number of Images per Class")
plt.xlabel("Classes")
plt.ylabel("Number of Images")
plt.show()
```

## Sample images from each class

```python
plt.figure(figsize=(12, 8))

for i, class_name in enumerate(class_names):
    class_path = os.path.join(DATASET_PATH, class_name)

    if os.path.isdir(class_path):
        images = [
            f for f in os.listdir(class_path)
            if f.lower().endswith(('.jpg', '.jpeg', '.png'))
        ]

        img_name = random.choice(images)
        img_path = os.path.join(class_path, img_name)

        with Image.open(img_path) as image:
            img = image.convert("RGB")

        plt.subplot(2, 3, i + 1)
        plt.imshow(img)
        plt.title(class_name)
        plt.axis("off")

plt.tight_layout()
plt.show()
```

## Check the original image sizes

```python
from collections import Counter

sizes = Counter()

for class_name in class_names:
    class_path = os.path.join(DATASET_PATH, class_name)

    if not os.path.isdir(class_path):
        continue

    for img_name in os.listdir(class_path):
        img_path = os.path.join(class_path, img_name)

        try:
            with Image.open(img_path) as img:
                img.convert("RGB").load()
                width, height = img.size
        except (OSError, ValueError):
            continue

        sizes[(height, width, 3)] += 1

print("Number of different image sizes:", len(sizes))

for size, count in sizes.most_common(10):
    print(size, ":", count, "images")
```

## Notes
- The classes are close in size but not perfectly balanced (see the bar chart).
- The images have different original sizes, so they must be resized in preprocessing.
- The clean folder has no unreadable or duplicated images (they were removed in 01).
