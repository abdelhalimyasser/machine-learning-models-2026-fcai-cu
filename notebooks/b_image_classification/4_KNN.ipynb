# 04 - K-Nearest Neighbors (KNN) Classifier

## Load the processed data

```python
import os
import pickle
import numpy as np

BASE_PATH = os.path.join("..", "..", "datasets", "b_classification_flowers")
PROCESSED_PATH = os.path.join(BASE_PATH, "processed")

data = np.load(os.path.join(PROCESSED_PATH, "flowers_split.npz"))

X_train = data["X_train"]
X_val = data["X_val"]
X_test = data["X_test"]
y_train = data["y_train"]
y_val = data["y_val"]
y_test = data["y_test"]

with open(os.path.join(PROCESSED_PATH, "classes.pkl"), "rb") as f:
    classes = pickle.load(f)

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
print("Classes:", classes)
```

## Try different K values

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

k_values = [3, 5, 7, 9]

best_k = None
best_accuracy = 0

for k in k_values:

    knn = KNeighborsClassifier(n_neighbors=k)

    knn.fit(X_train, y_train)

    y_val_pred = knn.predict(X_val)

    accuracy = accuracy_score(y_val, y_val_pred)

    print(f"K = {k}, Validation Accuracy = {accuracy:.4f}")

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_k = k

print("Best K:", best_k, "with Validation Accuracy = {:.4f}".format(best_accuracy))
```

## Train the best KNN model and save it

```python
knn = KNeighborsClassifier(n_neighbors=best_k)

knn.fit(X_train, y_train)

MODELS_PATH = os.path.join("..", "..", "models", "b_image_classification")
os.makedirs(MODELS_PATH, exist_ok=True)

with open(os.path.join(MODELS_PATH, "knn.pkl"), "wb") as f:
    pickle.dump(knn, f)

print("Best KNN model trained and saved successfully!")
```
