```python
import re
import pandas as pd
from bs4 import BeautifulSoup
```

```python
df = pd.read_csv("../../datasets/c_text_youm7/youm7_raw_data.csv")
```

```python
cols_2_drop = ["news_id", "source_site", "section_url",	"page_number",	"article_url", "image_url",	"publish_date_raw", "full_raw_html", "word_count_raw"]
df = df.drop(columns=cols_2_drop)
```

```python
df.head()
```

```python
def cast_to_string(text: object) -> str:
    if pd.isna(text):
        return ""
    return str(text)
```

```python
def remove_html_tags(text: str) -> str:
    text = BeautifulSoup(text, "html.parser").get_text(separator=" ")
    text = re.sub(r"http\S+|[^\w\s]", "", text)
    return text
```

```python
def remove_whitespace(text: str) -> str:
    return " ".join(text.split())
```

```python
def remove_empty_rows(df: pd.DataFrame) -> pd.DataFrame:
    df = df.dropna(subset=["title", "summary"])
    df = df[
        df["title"].str.strip().ne("") &
        df["summary"].str.strip().ne("")
    ]
    return df.reset_index(drop=True)
```

```python
def remove_duplicate_rows(df: pd.DataFrame) -> pd.DataFrame:
    return df.drop_duplicates(
        subset=["title", "summary"],
        keep="first"
    ).reset_index(drop=True)
```

```python
def clean_text_pipeline(text: object) -> str:
    text = cast_to_string(text)
    text = remove_html_tags(text)
    text = remove_whitespace(text)
    return text
```

```python
def clean_dataset_pipeline(df: pd.DataFrame) -> pd.DataFrame:
    df["title"] = df.pop("title_html").apply(clean_text_pipeline)
    df["summary"] = df.pop("summary_html").apply(clean_text_pipeline)

    df["category_label"] = (
        df["category_label"]
        .apply(cast_to_string)
        .apply(remove_whitespace)
    )

    df = remove_empty_rows(df)
    df = remove_duplicate_rows(df)

    return df
```

```python
df = clean_dataset_pipeline(df)
```
